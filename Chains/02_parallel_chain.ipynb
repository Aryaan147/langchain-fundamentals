```python
from langchain_groq import ChatGroq
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnableParallel
```

```python
model = ChatGroq(
    model="openai/gpt-oss-120b",
    temperature=0,
    max_tokens=512,
)
```

```python
parser = StrOutputParser()

template1 = PromptTemplate(
    template="make short notes and write all the key points for the following content:\n{content}",
    input_variables=["content"]
)
template2 = PromptTemplate(
    template="generate questions from the follwing topic:\n{content}",
    input_variables=["content"]
)
template3 = PromptTemplate(
    template="merge the following notes and questions into one single handbook\n\n\nnotes:\n{notes}\n\n\nquestions:\n{ques}",
    input_variables=["notes", "ques"]
)
```

```python
notes = template1 | model | parser
ques = template2 | model | parser

parallel_chain = RunnableParallel({
    "notes": notes,
    "ques": ques
})

merge = template3 | model | parser

chain = parallel_chain | merge
```

```python
content = """
Artificial Intelligence (AI) is a branch of computer science
that enables machines to perform tasks requiring human
intelligence, such as learning, problem-solving, and decision-making.

Machine Learning (ML) is a subset of AI that allows computers
to learn from data. Deep Learning uses neural networks to
learn complex patterns.

Natural Language Processing (NLP) helps computers understand
human language, while Computer Vision enables machines to
analyze images and videos.

AI is used in healthcare for disease diagnosis, education for
personalized learning, banking for fraud detection, and
transportation for self-driving vehicles.

Advantages of AI include automation, efficiency, and faster
decision-making. Disadvantages include privacy concerns,
job displacement, high costs, and ethical issues.
"""
```

```python
response = chain.invoke(content)
```
