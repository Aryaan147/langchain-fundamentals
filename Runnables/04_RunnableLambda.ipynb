```python
from langchain_groq import ChatGroq
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser 
from langchain_core.runnables import RunnableSequence, RunnableParallel, RunnablePassthrough, RunnableLambda
```

```python
model = ChatGroq(
    model="openai/gpt-oss-120b",
    temperature=0,
    max_tokens=512,
)
```

```python
template1 = PromptTemplate(
    template="Write an essay on topic: {topic}",
    input_variables=["topic"]
)

template2 = PromptTemplate(
    template="Word Count: {word_count}\n\n\nEssay:\n{essay}",
    input_variables=["word_count", "essay"]
)
```

```python
parallel = RunnableParallel({
    "essay": RunnablePassthrough(),
    "word_count": RunnableLambda(lambda x: len(x.split()))
})
chain = RunnableSequence(template1, model, StrOutputParser(), parallel, template2)
```

```python
response = chain.invoke({"topic":"sandard model"})
response
```
