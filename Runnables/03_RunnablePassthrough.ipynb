```python
from langchain_groq import ChatGroq
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser 
from langchain_core.runnables import RunnableSequence, RunnableParallel, RunnablePassthrough
```

```python
model = ChatGroq(
    model="openai/gpt-oss-120b",
    temperature=0,
    max_tokens=512,
)
```

```python
template1 = PromptTemplate(
    template="Write a detailed report on {topic}",
    input_variables=["topic"]
)

template2 = PromptTemplate(
    template="Write a 5 line summary on the following report.\n{text}",
    input_variables=["text"]
)

template3 = PromptTemplate(
    template="Report:\n{report}\n\n\n\nSummary:\n{summary}"
)
```

```python
prompt1 = template1.invoke({"topic":"Black hole"})
report = model.invoke(prompt1)
prompt2 = template2.invoke({"text":f"{report.content}"})
summary = model.invoke(prompt2)

summary
```

```python
parser = StrOutputParser()

report_gen = RunnableSequence(template1, model, parser)
parallel_chain = RunnableParallel({
    "report": RunnablePassthrough(),
    "summary": RunnableSequence(template2, model, parser)
})
chain = RunnableSequence(report_gen, parallel_chain, template3)
```

```python
chain
```

```python
chain.invoke({"topic":"black hole"})
```
