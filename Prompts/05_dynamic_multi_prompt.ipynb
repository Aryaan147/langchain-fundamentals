```python
from langchain_groq import ChatGroq
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder, load_prompt
from langchain_core.messages import HumanMessage, AIMessage
from dotenv import load_dotenv, find_dotenv

load_dotenv(find_dotenv())

model = ChatGroq(model="qwen/qwen3.8-27b")
```

```python
paper_input = "Attention Is All You Need"
style_input = "Beginner-friendly, intuitive, and technical"
length_input = "Detailed (approximately 1500 words)"
```

```python
template = ChatPromptTemplate.from_messages([
    (
        "system",
        """
You are an expert research paper explainer.

Explain research papers according to the user's requested style and length.

If certain information is not available in the paper, respond with:
"Insufficient information available" instead of guessing.
"""
    ),
    MessagesPlaceholder(variable_name="chat_history"),
    (
        "human",
        """
Please summarize the research paper titled "{paper_input}"
with the following specifications:

Explanation Style: {style_input}
Explanation Length: {length_input}

1. Mathematical Details:
    - Include relevant mathematical equations if present in the paper.
    - Explain mathematical concepts using simple, intuitive code snippets where applicable.

2. Analogies:
    - Use relatable analogies to simplify complex ideas.

Ensure the summary is clear, accurate, and aligned with the provided style and length.
"""
    )
])
```

```python
chat_history = []
```

```python
prompt = template.invoke({
    "paper_input": paper_input,
    "style_input": style_input,
    "length_input": length_input,
    "chat_history": chat_history
})
```

```python
response = model.invoke(prompt)
```

```python
chat_history.append(HumanMessage(content=f"""
Summarize the paper titled "{paper_input}".
Style: {style_input}
Length: {length_input}
"""))

chat_history.append(AIMessage(content=response.content))
```

```python
print(response.content)

while True:
    query = input()

    if(query.lower == "bye"):
        break

    chat_history.append(HumanMessage(content=query))

    prompt = template.invoke({
        "paper_input": paper_input,
        "style_input": style_input,
        "length_input": length_input,
        "chat_history": chat_history[:-1]
    })

    response = model.invoke(prompt)

    print(response.content)

    chat_history.append(AIMessage(content=response.content))
```
