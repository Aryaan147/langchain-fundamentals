# Using System, Human, and AI Messages
A simple LangChain example demonstrating how to send system instructions and a human query to a Groq chat model, then store the AI response in the message history.

## 1. Import libraries

```python
from langchain_core.messages import AIMessage, HumanMessage, SystemMessage
from langchain_groq import ChatGroq
```

## 2. Initialize the model

```python
model = ChatGroq(model="qwen/qwen3.8-27b")
```

## 3. Create the message list
- `SystemMessage` defines the assistant's behavior.
- `HumanMessage` contains the user's question.

```python
messages = [
    SystemMessage(
        content="You are a helpful technical AI assistant who helps beginners understand how technologies work and where they are used."
    ),
    HumanMessage(content="What is LangChain?")
]
```

## 4. Invoke the model

```python
result = model.invoke(messages)

print(result.content)
```

## 5. Append the AI response to the message list
Store the assistant's response as an `AIMessage` so the complete conversation can be reused in subsequent model calls.

```python
messages.append(AIMessage(content=result.content))

print(messages)
```
