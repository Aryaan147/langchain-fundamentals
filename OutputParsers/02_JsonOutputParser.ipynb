```python
from langchain_huggingface import ChatHuggingFace, HuggingFacePipeline, HuggingFaceEndpoint
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import JsonOutputParser 
import os

os.environ["HF_HOME"] = "/Users/aryan/Downloads/hf_cache"
```

```python
llm = HuggingFacePipeline.from_model_id(
    model_id="TinyLlama/TinyLlama-1.1B-Chat-v1.0",
    task = "text-generation",
    pipeline_kwargs={
        "temperature":0.3
    }
)

# llm = HuggingFaceEndpoint(
#     repo_id="google/gemma-2-2b-it",
#     task = "text-generation"
# )

 
model = ChatHuggingFace(llm=llm)
```

```python
parser = JsonOutputParser()
```

```python
template = PromptTemplate(
    template="Give me the name, age and city of a fictional character\n{format_instruction}",
    partial_variables={"format_instruction": parser.get_format_instructions()}   # partial variable bcz place holder is replaced at compile time (before run time)
)

prompt = template.format()
```

```python
# response = model.invoke(prompt)
# final_response = parser.parse(response.content)
# print(final_response)
```

```python
chain = template | model | parser

response = chain.invoke({})
print(response)
```
